### One parent One Child

```python
#parent
class Phone:
    category="Electronics" #this is class attribute
    

    #constructor
    def __init__(self, model,battery,camera,charge):
        self.model=model #these are Instance Attribute
        self.battery=int(battery)
        self.camera=camera
        self.charge=charge

    #Methods
    def capture(self, number_of_photoes):
        self.charge-=(0.25*number_of_photoes)

    def see_charge(self):
        return self.charge

#child
class Smartphone(Phone): #inherited from Phone

    category = "Smart Phone" #This overrite 'category' in parent. This is class attribute of Smartphone now.

    #constructor
    def __init__(self,model,battery,camera,charge):
        super().__init__(model,battery,camera,charge) #Passed parent parameters to the parent class
        #super() used to access parent Class Attributes not the instance attributes but can not modify them also used to access the parent class methods. Here we accessed the parent constructor method

    def fast_charge(self,minute):
        self.charge+=(1.75*minute)
        if(self.charge>=100):
            self.charge=100
            print("Fully Charged")


#object create
pro = Smartphone("x",5000,50,50)

print(pro.category)

pro.fast_charge(24)

print(pro.charge)

```

### Multiple parent One child

```python
#Parent-1
class Camera:
    def __init__(self, megapixels): #constructor function
        self.megapixels = megapixels

    def take_photo(self):
        print(f"Taking a photo with a {self.megapixels} MP camera.")

#parent-2

class MusicPlayer:
    def __init__(self, storage):
        self.storage = storage

    def play_music(self):
        print(f"Playing music using {self.storage} GB of storage.")


#child
class Smartphone(Camera, MusicPlayer):
    def __init__(self, megapixels, storage, brand):
        Camera.__init__(self, megapixels) #super().__init__ doesnot need the 'self' parameter but here we must put it. super() handles the method binding, so you don't need to pass 'self' explicitly.
        MusicPlayer.__init__(self, storage)
        self.brand = brand

    def make_call(self):
        print(f"{self.brand} smartphone is making a call.")


phone1 = Smartphone(50,128,"Samsung")

print(f"{phone1.megapixels}, {phone1.storage}, {phone1.brand}")

phone1.take_photo()
phone1.play_music()
phone1.make_call()
```
