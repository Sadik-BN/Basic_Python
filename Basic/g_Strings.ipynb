### String methods

```python
str1="Sadik Khan Adnan"

#all small letter
sml_str1= str1.lower()
print(sml_str1)

#all capital letter
cap_str1= str1.upper()
print(cap_str1)

#Only first character Capital
str2= str1.capitalize()
print(str2)

#length
print(len(str1))

#Find a word (Case sensitive) (return true/false)
print("Khan" in str1)
print("khan" in str1)

#find starting index of a substring (if not found return -1, for multiple findings returns first finding)
indx = str1.find("an")
print(indx)

# to get amonmultiple findings returns last finding
indx = str1.rfind("an")
print(indx)

# Count same substring in a string
cnt = str1.count("an")
print(cnt)

#Find and replace
changed_str1=str1.replace("an","AN")
print(changed_str1)
```

### Splitting and Joining

```python
prompt = "Tell me about python ."

words= prompt.split() #split based on white space by default
print(words)

date="26-09-2026"
nums= date.split("-") #split based on custom character
print(nums)

words= prompt.split(" ",2) # number of times the string will split based on the character (whitespace for this case)
print(words)

date= "/".join(nums) #Join a Iterable container like list with a ciustom character
print(date)
```
