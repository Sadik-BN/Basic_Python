```python
#Normal if else

raining = True

if raining == True:
    print("Caution")
    print("Wear Raincoat")
else:
    print("It is a sunny day")

#ladder
money= 500
if money>=500 and raining == False:
    print("Go outside")
elif money>=400:
    print("Order food")
elif money>=300 and raining==False:
    print("Go for food cart")
else:
    print("Save money")


#Inline
n=48

print("Even" if n%2==0 else "Odd")


#Inline Chain
marks=57

print("A+" if marks>=80 else "A" if marks>=75 else "A-" if marks>=70 else "B" if marks>=60 else "F")
```
