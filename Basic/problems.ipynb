```python
inpt = input()

input_list=inpt.split() #splitted the input based on whitespace by default (to split based on ',' , '.' , '#' , '*' etc we put them inside 'split(",")' function) and stored in a list as string (list array type thing inside 3rd bracket ['10','20','30','40'])

m= int(input_list[0]) #values in every index stored as string in list so typecasted to int
n=int(input_list[1])  

print((m%10)+(n%10))
```

```python
inpt_string = input()
input_list= list(map(int,inpt_string.split()))

min = input_list[0]
mx = input_list[0]

for x in input_list:
    if x<min:
        min=x
    if x>mx:
        mx=x

print(min,mx)
    
```

```python
n = int(input())

arr = list(map(int,input().split()))

ev_cnt=0
odd_cnt=0
pos_cnt=0
neg_cnt=0

for x in arr:
    if x%2==0 :
        ev_cnt+=1
    if x%2!=0:
        odd_cnt+=1
    if x>0:
        pos_cnt+=1
    if x<0:
        neg_cnt+=1

print("Even:", ev_cnt)
print("Odd:", odd_cnt)
print("Positive:", pos_cnt)
print("Negative:", neg_cnt)
```
