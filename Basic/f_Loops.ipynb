### For Loop

```python
#Type-1

for i in range(10):
    print(i,end=" ") #prints 0 to 9
print()


#Type-2

for i in range(2,8):
    print(i,end=" ") #prints 2 to 7
print()


#Type-3 

for i in range(10,0,-1):
    print(i,end=" ") #prints (10 to 1) step -1
print()


#Signature 'range(start,stop,step)'


#Type-4(Container Iteration)

fruits = ["Apple","Orange","Grape","Banana","Dates"]

for fruit in fruits:
    print(fruit,end=" ")
```

### While Loop

```python
count=0

while count < 10:
    count+=1
    print(count,"Hello World",sep=". ",end=".\n")
    
```

### Both Loops contain <strong> 'continue' </strong> and <strong>'break'</strong> statements same as other programming languages
