### Polymorphism using method overriding

```python
class Camera:
    def __init__(self,megapixels):
        self.megapixels=megapixels

    def capture(self):
        print("photo captured")



class Phone(Camera):
    def __init__(self,megapixels,brand):
        Camera.__init__(self,megapixels)
        self.brand= brand

    def capture(self):
        print(f"Photo captured with {self.brand} phone")  #this will override the parent class's capture() function


phone=Phone(50,"Motorola")

phone.capture()
```
