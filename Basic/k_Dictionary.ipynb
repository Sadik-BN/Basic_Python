
# Dictionary
### Data stored in key,value pair
### unordered
### mutable

### \# keys can be any data type except any mutable data type like list
### \# values can be anything

```python
#Declaration 

# empty disctionary

dic={}
print(type(dic),dic)


#declare
dic={
    "name":"Sadik",
     "age":22,
     "address":"Dhaka",
     "numbers":[10,20,30,40]
    }
print(type(dic),dic)
```

```python
#Access
print(dic["name"])
print(dic["numbers"])
#print(dic["phone"]) #if not found gives key error

#Using Function
print(dic.get("age"))
print(dic.get("phone")) #if not found gives no error. returns None by default.
#to return a custom value 
print(dic.get("birth_date",0))
```

```python
#Assign Values

dic["age"]=23 #this will update the age

dic["phone"]="0171234567" #phone is not present so this will add the phone
#using function
dic.update({"birth_date":"3,Dec"})

print(dic)

#Duplicate handling
dic={
    "name":"Sadik",
     "age":22,
     "address":"Dhaka",
     "numbers":[10,20,30,40],
     "age":30
    }
print(dic)
# keeps the latest value of duplicate key
```

```python
#Deleting
del dic["address"]
print(dic)
```

```python
#methods
dic={
    "name":"Sadik",
     "age":22,
     "address":"Dhaka",
     "numbers":[10,20,30,40],
    }
#Coppying
dic2= dic.copy()
print(dic2)
```

### \# keys can be any data type except any mutable data type like list
### \# values can be anything

```python
#dic={[10,30,20]:"a"} this will give error cause list is mutable 

#but
dic={
    (10,20,30):"a",
    "b":[10,20,30],
    "c":{
        "name":"sadik",
        "dept":"cse"
    }
}
print(dic)
#This declaration is valid 
print(dic["c"])
print(dic["c"]["name"])
print(dic.get("c").get("dept"))
```

```python
#View methods

dic={
    "name":"Sadik",
     "age":22,
     "address":"Dhaka",
     "numbers":[10,20,30,40],
    }

#Extract the keys dynamicly as list in a data structure called dict_keys 

dic_keys= dic.keys()
print(type(dic_keys),dic_keys)

#If new key added no need to do dic.keys() again. 'dic_keys' will be automtically updated
dic["hometown"]="Sirajganj"
print(dic_keys)


#Extract the values dynamicly as list in a data structure called dict_values
dic_values = dic.values()
print(dic_values)
#also here if a new key:value added no need to do dic.values() again. 'dic_values' will be automtically updated



#extract both key:value pair dynamically as tuples in a list in a data structure called dict_items

dic_items= dic.items()
print(dic_items)
print("\n")



#Iteration

#key iteration
for key in dic.keys():
    print(key,end=" ")
print("\n")    

#value iteration
for value in dic.values():
    print(value,end=" ")
print("\n")  

#Pair iteration
for key,value in dic.items():
    print(f"{key} : {value}")
```

### Dictionary Comprehension

```python
# a_dictionary= {<expression(key:value)> for <ierables> if<conditions>}

square={x:x**2 for x in range(1,11)} #this will take numbers from (1-10) as key and its square as value
print(square)

evenSquares={x:x**2 for x in range(1,11) if x%2==0} #this will take even numbers between (1-10) as key and its square as value
print(evenSquares)


#Multiple data structure iteration and comprehension in dictionary
co_ordinates=[(23.811056, 90.407608),(22.341900, 91.815536),(24.886436, 91.880722)]
location=['Dhaka','Chattogram','Sylhet']

exact_location= {cord:loc for cord,loc in zip(co_ordinates,location)}
print(exact_location)
```
