# File

### File Read

```python
#File Read Method-1

file = open("./sample_data/sample.txt","r") #open a file in read mode
print(type(file))

content = file.read() #read the whole file as string
print(type(content))
print(content) #print the string

file.seek(0) #For first read() a pointer will go through the file like a coursor and stop at the end. To attempt another read we need to set the pointer/cursor at position 0 of by using seek(0)


#Get the lines with trailing '\n' from content string
content = file.readlines() #read the file line by line. Returns the lines as string in a list. The lines will contain '\n' too from the file
print(content) #print the list of lines

file.seek(0)

#get the list of lines with '\n' which is 'content' and make them without '\n'
content = list(map(lambda line:line.strip(),content))#map each line from the list 'content' then strip them (remove leading and trailint spaces,'\n','\t' and returns rest of the string) and created a list, stored in same 'content'
print(content)




print(file.closed) #to check file is clossed or not

#We must close the file after work

file.close()#Used To close the file
print(file.closed)#new status
```

```python
#File read method-2 (best practice)

with open("./sample_data/sample.txt","r") as file:

  #we can print lines directly from file without reading. Using range based loop from file (This will print '\n' too)
  for line in file:
    print(line,end="") #ended with "" still will have new line from the file

  file.seek(0)

  #To print without new line '\n' of the file
  for line in file:
    l = line.strip() #stores the line in 'l' without new line '\n'
    print(f"{l}",end=". ")


#method-2 auto closes the file after executing the block of code no need to close
print(f"\n{file.closed}") #Status check
```

### File Write

```python
#Writing in a new file
with open("./sample_data/test.txt","w") as file: #this will open a file in write mode if file not exist this will create one
  file.write("Hello\n")
  file.write("Testing file write") #this will write in file


#writing in existing file without overwriting previous data
with open("./sample_data/sample.txt","a") as file:
  file.write("\nPython is great\n")
  file.write("Anything can be done with it")
```
