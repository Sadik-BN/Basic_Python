### List Declaration

```python
numbers=[10,20,32,1,43,9]

print(type(numbers))

print(numbers[2])
numbers[2]=49
print(numbers[2])
#List is a mutable structure


#reverse indexing
print(numbers[-1])
print(numbers[-2])
print(numbers[-3])

mixed_type_list = [10,3.1416,'Apple']
print(mixed_type_list)

list_inside_list =[[10,20,30],[32,65,13],[98,97,100]]
print(list_inside_list[1][2])
```

### Slicing

```python
# sliced_list_name[start:end:step]

new_list = numbers[1:4:1]
print(new_list)

new_list= numbers[3:0:-1]
print(new_list)
```

### Modification

```python
# Adding an element at last

numbers=[0,12,2,87,23,6]
numbers.append(999)
print(numbers)

#insert in any position
numbers.insert(2,0)
print(numbers)

#pop last element
numbers.pop()
print(numbers)

#pop from any position
numbers.pop(3)
print(numbers)

#numbers.pop() also returns the popped value if printed or stored


#remove any element only first occurance
numbers.remove(0)
print(numbers)
```

### Sorting and Reversing

```python
numbers=[0,12,2,87,23,6]
#ascending sort
numbers.sort()
print(numbers)

#descending sort
numbers.sort(reverse=True)
print(numbers)


#reverse the list 
numbers=[0,12,2,87,23,6]
numbers.reverse()
print(numbers)
```

### Stack

```python
stack =[]

#pussh in stack
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

#getting top and pop
print(stack[-1]) #top
stack.pop() #pop
print(stack[-1])
stack.pop()
print(stack[-1])
stack.pop()
print(stack[-1])
stack.pop()
print(stack[-1])
```

### Queue

```python
queue = []

#push in
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

#getting front and pop
print(queue[0]) #front
queue.pop(0) #pop
print(queue[0])
queue.pop(0)
print(queue[0])
queue.pop(0)
print(queue[0])
queue.pop(0)
print(queue[0])
```

### List comprehension

```python
# create a list of even numbers from 0 to 20 

even = [x for x in range(0,21) if x%2==0]
print(even)

#we can add,subtract,multiple,devide these filtered even numbers too. for example:-
double_of_evens= [x*2 for x in range(0,21) if x%2==0]
print(double_of_evens)


#list of strings
fruits=['apple','mango','lichi','guava']

capitalized_fruits = [fruit.capitalize() for fruit in fruits]
print(capitalized_fruits)

#So, any_list = [<expression_that_will_be_stored> <for_loop_iteration> <Condition_of_storing>]
#Not only list any iterable structure can be comprehensed into list
```
