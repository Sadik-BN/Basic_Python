Operators

```python
#arithmatic 
x=10; y=3

sum= x+y
sub= x-y
mult= x*y
div = round(x/y,2) #always stores float by default. 'round()' used to show nubers after decimal point
remainder = x%y
print(sum,sub,mult,div,remainder)
print(type(sum),type(sub),type(mult),type(div),type(remainder))
```

Comparison Operators: >, <, >=, <=, ==, !=

Logical Operators: and, or
