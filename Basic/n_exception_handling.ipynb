```python
#Full Version of error handling

try:
  file = open("./sample_data/sample1.txt","r")
except Exception as e: #show the error
  print(e)
else: #If no error then else block will be executed
  content = file.read()
  print(content)
  file.close()
finally: #error occurs or not this block will be executed at last
  print("End of reading")
```
