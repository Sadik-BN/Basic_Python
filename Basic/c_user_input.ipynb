Taking Input in python:
 We use 'input("Input prompt" or keep empty)' function to take input which always returns string. No matter User Entered any number or string. 

```python
age = input("Enter your age:")
print("You are",age,"years old\nType of entered age:",type(age))

age = int(age)

print("After Typecasting",type(age))
```

### Row of numbers

```python
row_of_numbers= input()

numbers = row_of_numbers.split() #this will split the string of 'row_of_numbers' based on whitespace and put them in the 'numbrs' list 

#then we can typecast each number
a= int(numbers[0])
b= int(numbers[1])
c= int(numbers[2])
print(type(a),a,type(b),b,type(c),c)

#Or all numbers together
numbers=[int(number) for number in numbers]

print(numbers,type(numbers[1]))
```
