# Python OOP: Abstraction

## 1. What is Abstraction?

**Abstraction** means exposing the essential features of something while hiding unnecessary implementation details.

For example, when you use a coffee machine, you press a button to make coffee. You do not need to know every internal step the machine performs.

In programming, abstraction lets us define **what an object should do** without requiring every user of the object to know **how it does it**.

### Simple idea
- **Public interface:** the methods other code is expected to use.
- **Internal implementation:** the details behind those methods.


## 2. A Simple Example

A `CoffeeMachine` exposes one simple method, `make_coffee()`. The other methods represent internal steps.

Python does not automatically make these methods private; the double underscore triggers name mangling. This example uses it to signal that the steps are implementation details.

````python
class CoffeeMachine:
    def make_coffee(self):
        self.__heat_water()
        self.__brew_coffee()
        print("Coffee is ready!")

    def __heat_water(self):
        print("Heating water...")

    def __brew_coffee(self):
        print("Brewing coffee...")


machine = CoffeeMachine()
machine.make_coffee()
````

### What to notice

You only need to call `machine.make_coffee()` to use the machine. The method handles the internal steps.

**Important:** In Python, double-underscore methods are not truly inaccessible. Python changes their names through *name mangling* to reduce accidental access and naming conflicts; this is not a security feature.


## 3. Abstract Classes and `@abstractmethod`

Python provides the `abc` module for defining formal abstract classes.

- `ABC` is a base class used to create an abstract base class.
- `@abstractmethod` marks a method that concrete subclasses must implement.
- You cannot instantiate a class while it still has unimplemented abstract methods.
- Different subclasses can implement the same method in different ways.


````python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass


class Dog(Animal):

    def make_sound(self):
        print("Woof!")


class Cat(Animal):

    def make_sound(self):
        print("Meow!")


dog = Dog()
cat = Cat()

dog.make_sound()
cat.make_sound()
````

### Why does `Animal` inherit from `ABC`?

`Animal(ABC)` makes `Animal` an abstract base class. The `@abstractmethod` decorator marks `make_sound()` as required.

`pass` is simply a placeholder: it means the method has no implementation here. Each concrete child class supplies its own behavior.


````python
class Fish(Animal):
    pass


fish = Fish()  # Raises TypeError: make_sound() is not implemented
````

The `Fish` example raises `TypeError` when you try to create an object because `Fish` has not implemented `make_sound()`.

You also cannot create an object directly from `Animal` while its abstract method remains unimplemented:

```python
animal = Animal()  # TypeError
```


## 4. A Practical Example: Payment Methods

Different payment methods work differently, but an application can rely on a shared `pay()` interface.

````python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CardPayment(Payment):

    def pay(self, amount):
        print(f"Paid {amount} using a card.")


class CashPayment(Payment):

    def pay(self, amount):
        print(f"Paid {amount} in cash.")


payments = [CardPayment(), CashPayment()]

for payment in payments:
    payment.pay(500)
````

The code using `Payment` can call `pay(amount)` without needing to know the internal details of each payment type.

This example demonstrates a **common interface** with different implementations. In a real payment system, actual payment processing would also need proper validation, error handling, and secure handling of payment data.


## 5. Abstraction vs. Encapsulation

| Abstraction | Encapsulation |
|---|---|
| Hides unnecessary implementation complexity behind a clear interface. | Bundles data and methods and controls how an object's state is accessed or changed. |
| Focuses on **what an object does**. | Focuses on **how data and internal state are managed**. |
| Example: call `make_coffee()` without knowing the brewing steps. | Example: validate a bank-account withdrawal before changing its balance. |

They often work together, but they solve different problems.


## 6. Key Points to Remember

1. Abstraction keeps the user-facing interface simple.
2. An abstract class can define a common structure for related classes.
3. `ABC` and `@abstractmethod` are provided by Python's `abc` module.
4. A concrete subclass must implement all inherited abstract methods before it can be instantiated.
5. Abstract methods can contain code, but subclasses still need to override them to satisfy the abstract requirement.
6. Abstraction is a design technique, not a security boundary.
