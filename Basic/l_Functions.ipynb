### Types of arguments

```python
#Positional Arguments

def multiply(a,b):
    return a*b

print(multiply(10,5)) #we give arguments by position. Here 10->a and 5->b


#for Unlimited arguments/ Variable sized arguments
def func(*args): #user can pass so many arguments in this fuction. All will be stored in a tuple named args. We can use the arguments by extracting from the tuple
    print(args)

func(10,20,30,40,50)





#Keyword Arguments
#we can pass the arguments by parameter name
def func3(a,b,c):
    print(f"A={a},B={b},C={c}")

func3(b=20,c=30,a=10)


#For unlimited keyword arguments/ VAriable sized keyword arguments
def func4(**keyword_args): #user can pass so many keyword arguments in this function. All will be stored in a dictionary named keyword_args. We can use them extracting from dictionary
    print(keyword_args)

func4(sadik=22,z=1,a=20,d=99)





#mixed arguments passing
func3(22,c=31,b=12) #if we pass a positional argument in a position that argument cant be sent as keyword argument again.


```

### Iterator


```python
# Iterator is an object that returns elements one at a time from a sequence/data stream and remmebersits position between calls

#Set has no index so we do iteration here

S = {10,20,30,40,50}

S_iterator = iter(S)

print(S_iterator)
print(next(S_iterator))
print(next(S_iterator))
print(next(S_iterator))
print(next(S_iterator))
print(next(S_iterator))#this is the last iteration after this no value can be found so it will rise a StopIteration error

```

### Generator

```python
#Generator is a function that behaves like an iterator(it can used in a for loop)
# pause a function, return a value, the resumes
# Uses 'yield' instead of return
# Iterate without loading everything into memory (ex- reading large files)

def count_to(n):
    numbers=[]
    count = 1
    while count <=n:
        numbers.append(count)
        count+=1
    return numbers

for i in count_to(10): #This is normal function. If the number is huge like 1000000000000000 this will create a huge list of numbers and return here where we will get a memory error
    print(i)




#Here is GENERATOR
def count_to(n):
    count = 1
    while count <=n:
        yield count
        count+=1

#This function will return one number and then continue to next (if we ask for by iteration) like pause then play again not a whole counting list or anything. We can iterate every returns with for loop or next() function

print(type(count_to(10)))

for i in count_to(10): #Here for 10000000000 this will run for so long but will not give any memory error cause it taking space of only one number and returning it at each iteration where normal funtion will iterate over a huge list and return it
    print(i)






#Above we were returning one by one value from huge data. We can also return small chunks from huge data by using a generator

lst =[x for x in range(51)]

def genrator(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]

x=genrator(5,lst)

#Iteration
#Manual iteration
print("Manual Iteration Starts Here")
print(x)
print(next(x))
print(next(x))


#For loop. For loop will starts to iterate just after manual iteration finished it wont start from begining. Cause once iterator/generator starts to iterate these can't re-iterate from begining. Once Iterated is Iterated Forever

print("For loop iteration starts here: ")
for chunk in x:
    print(chunk)


```

### Special Functions

```python
#lambda Function

lambda a,b: a+b #this function used inside apply(),map() 


#___.apply() Function -> needs a lambda funnction to work. This function can be used only with some specific data structures

#Example:-
import pandas as pd

data_frame = pd.DataFrame({'Temperature (F)':[32,212,98.6,68]}) # This will create a dataframe which is like table with a column named 'Temperature (F)' with the temperature values under it

#We will calculate the temperature in celcius and add them in another column named 'Temperature (C)' in this data frame

data_frame['Temperature (C)'] = data_frame['Temperature (F)'].apply(lambda x: 5*((x-32)/9)) #This is how apply and lambda work together

print(data_frame)



#___.map() Function

data_frame2 = pd.DataFrame({'A':[16,169,121,64,81],'B':[10,20,30,40,50]})



data_frame2['A'] = data_frame2['A'].map(lambda x: x**0.5) # this can be done with apply() too

print(data_frame2)
```
