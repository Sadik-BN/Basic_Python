### Class and Object

```python
class Phone:
    category="Electronics" #this is class attribute
    

    #constructor
    def __init__(self, model,battery,camera,charge):
        self.model=model #these are Instance Attribute
        self.battery=int(battery)
        self.camera=camera
        self.charge=charge

    #Methods
    def capture(self, number_of_photoes):
        self.charge-=(0.25*number_of_photoes)

    def see_charge(self):
        return self.charge


#Creating Object
i_phone=Phone("iphone-17 pro max",3500,120,76)
blueberry = Phone("b27",5000,80,55)
motorola = Phone("G64-5G",6000,60,97)

print(type(i_phone))
print(i_phone.category)

print(f"Motorola Phone -> {motorola.category},{motorola.model},{motorola.battery} mAh,{motorola.camera}mp")

motorola.capture(30)
print(motorola.see_charge())


print(i_phone.__dict__) #To see an object as a dictionary
```
