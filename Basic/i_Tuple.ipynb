# Tuples
### Immutable collection of data

```python
# Delaration
tup=(10,20,30)

print(type(tup))

mixed_tup = (10,3.33,"sadik",True)
print(mixed_tup)

#type of single valued tuple is not tuple 
tup1= (12)
print(type(tup1)) #its the type of the value, int

#to make it tuple
tup1=(12,)
print(type(tup1))
```

```python
# Tuple to list and vice versa
tup=(1,2,3,4)
lis= list(tup)
print(lis)


lis2=[12,4,53,1]
tup2= tuple(lis2)
print(tup2)
```

```python
#Access
tup=(10,30,20,10.5,"Sadik",False)
print(tup[4])


#slicing 
#sliced_tuple=[start:end:step]

new_tup=tup[2:5:1]
print(new_tup)
```

```python
#Tuple metod
tup=(10,20,30,40,10,20,30)

#count ocuurance
print(tup.count(10))

#First found index
print(tup.index(40))
```
